## 3.Homogenización y fusión de bases de datos
En este proyecto, se decidió que la estructura indicada para homogeneizar las bases de datos respetaría el orden donde los renglones contienen el comportamiento de todos los parámetros de monitoreo en una hora específica del día. 

```python
import pandas as pd
```

```python
# Crear carpeta para los archivos procesados
carpeta_salida = './Archivos/Datos/Homogeneizados'
```

### Periodo 1996-2011
Para homogeneizar los datos monitoreo de la calidad del aire desde **1996** a **2011** se deben tener en cuenta los siguientes puntos: 
1. Se desea invertir la estructura de las bases de datos al transformar las horas en renglones y los parámetros en columnas.
2. La hora 01 del día corresponderá a las `01:00:00` horas, mientras que la hora 24 se considerará las `00:00:00` del día posterior.
3. El archivo `datos_2011` tiene una columna adicional llamada `INVALIDA` que debe ser removida.
4. También, el archivo `datos_2011` contiene datos del 31 de diciembre de 2010 que deberán ser eliminados.

La metodología de homogenización se presenta a continuación:

1. Uso del método `.melt()` en la biblioteca pandas para reestructurar los datos al transformar columnas en filas. En este caso, el método convierte las columnas de horas (`HORA01`, `HORA02`, etc.) en filas bajo una columna llamada `HORA`. Los parámetros claves del método son: 
- `id_vars`: Las columnas que se desean conservar o no se necesitan invertir a renglones (`CLAVE_EST`, `FECHA` y `PARAMETRO`).
- `value_vars`: Las columnas que se beben convertir en filas (`HORA 01` a `HORA 24`). 
- `var_name`: El nombre de la nueva columna que contendrá los nombres de las columnas invertidas (`HORA`).
- `value_name`: El nombre de una nueva columna que contendrá los valores de las columnas invertidas (`VALOR`).

2. Se extrajeron los números de las horas mediante `.str.replace()` y se convirtieron en valores enteros con `.astype(int)`. Posteriormente, esos valores numéricos (como 0, 1, 2, etc.) se trasformaron en un formato de hora (`hh:mm:ss`) mediante `.to_timedelta()`.

3. Cada elemento de la columna `PARAMETRO` (como CO, NO2, etc.) se convirtió en una columna única, con sus valores medidos correspondientes, mediante `.pivot_table()` con la siguiente configuración de parámetros del método: 
- `index`: Define los renglones (`CLAVE`, `FECHA` y `HORA`).
- `columns`: Define qué elementos se convierten en columnas (`PARAMETRO`).
- `values`: Especifica qué datos se agregan y organizan (`VALOR`).
- `aggfunc`: Aplica una función de agregación (`mean` evita duplicados)

4. Respecto al archivo `datos_2011`, se eliminó la columna `INVALIDA` con el método `.drop()`.


```python
# Procesar los archivos de 1996 a 2011, uno por uno
for año in range(1996,2011+1):
    print(f"Procesando: datos_{año}.xlsx")  # Mensaje de seguimiento

    # Cargar archivo Excel
    df = pd.read_excel(f'./Archivos/Datos/datos_{año}.xlsx')

    # Eliminar la columna "INVALIDA" si existe
    if 'INVALIDA' in df.columns:
        df = df.drop(columns=['INVALIDA'])

    # Aplicar homogeneización
    df_melted = df.melt(
        id_vars=['CLAVE_EST', 'FECHA', 'PARAMETRO'],
        value_vars=[col for col in df.columns if col.startswith('HORA')],
        var_name='HORA',
        value_name='VALOR'
    )

    # Convertir la columna HORA a número entero eliminando el prefijo "HORA"
    df_melted['HORA'] = df_melted['HORA'].str.replace('HORA', '').astype(int)

    # Manejo del caso especial: Si HORA es 24, moverlo al siguiente día y asignar 00:00:00
    df_melted.loc[df_melted['HORA'] == 24, 'FECHA'] = pd.to_datetime(df_melted['FECHA']) + pd.to_timedelta(1, unit='d')
    df_melted.loc[df_melted['HORA'] == 24, 'HORA'] = 0

    # Convertir HORA en formato timedelta para eliminar "0 days"
    df_melted['HORA'] = pd.to_timedelta(df_melted['HORA'], unit='h').astype(str).str.replace("0 days ", "")

    # Convertir la columna VALOR a formato numérico para evitar errores de conversión
    df_melted['VALOR'] = pd.to_numeric(df_melted['VALOR'], errors='coerce')

    # Pivotear para organizar por parámetros
    df_pivoted = df_melted.pivot_table(
        index=['CLAVE_EST', 'FECHA', 'HORA'],
        columns='PARAMETRO',
        values='VALOR',
        aggfunc='mean'
    ).reset_index()

    # Homogeneizar nombres de columnas
    columnas_homogeneas = {'O': 'O3'}
    df_pivoted.rename(columns=columnas_homogeneas, inplace=True)

    # Lista de columnas a eliminar
    columns_to_drop = ['STW', 'STWD', 'FILT', 'FLAM', 'THC', 'MHC']
    # Eliminar columnas solo si existen en el DataFrame
    df_pivoted = df_pivoted.drop(columns=[col for col in columns_to_drop if col in df_pivoted.columns])

    # Eliminar columnas con nombres duplicados en el índice
    df = df.loc[:, ~df.columns.duplicated()]
    
    # Guardar cada archivo procesado en la carpeta
    df_pivoted.to_excel(f'{carpeta_salida}/datos_{año}_homogeneizado.xlsx', index=False)

    print(f"Paso completado: Guardado como {carpeta_salida}/datos_{año}_homogeneizado.xlsx")
```

Este proceso de homogenización se realizó por cada base de datos, guardando los archivos procesados con el nombre de `datos_{año}_homogeneizado`. Estos nuevos archivos se concatenaron mediante el método `.append.()` y `.concat()`. El dataframe final comprende datos desde 1996 a 2011

```python
# Inicialización de la Lista
archivos = []

# Asignación de los Archivos a la Lista
for año in range(1996, 2011+1):
    df = pd.read_excel(f'./Archivos/Datos/Homogeneizados/datos_{año}_homogeneizado.xlsx')

    # **Filtrar registros que correspondan al año del archivo**
    df = df[df['FECHA'].dt.year == año]

    # Eliminar columnas con nombres duplicados en el índice
    df = df.loc[:, ~df.columns.duplicated()]

    archivos.append(df)  # Agregar el DataFrame filtrado a la lista
    
# Concatenación de los Elementos de la Lista a un DataFrame
df_96_11 = pd.concat(archivos, ignore_index = True)

# Guardar el DataFrame final como CSV
ruta_guardado = "./Archivos/Datos/Homogeneizados/datos_1996_2011_homogeneizado.csv"
df_96_11.to_csv(ruta_guardado, index=False, encoding="utf-8")

# Mostrar el tipo de datos contenidos en el Dataframe
print('Tipos de Datos en el DataFrame:\n', df_96_11.dtypes)

# Impresión de los Resultados
display(df_96_11)
```

### Periodo 2012-2018 y 2021
Para homogeneizar los datos monitoreo de la calidad del aire desde *2012* a *2018*, así como los datos de *2021*, se deben tener en cuenta los siguientes puntos: 
1. A diferencia de las bases de datos del periodo anterior, el presente periodo de datos tiene un formato donde los parámetros y contaminantes son columnas y las horas de monitoreo son renglones. Sin embargo, hay una diversidad de nombres asociados a los parámetros. Esta problemática de nomenclaturas de referencia también se repite en con el título de las columnas relacionadas con la fecha y hora de monitoreo.
2. Es necesario recolectar la información desde diferentes hojas de las bases de datos originales (hojas de cálculo de Excel). Cada hoja recopila el monitoreo de una estación por específico, con la problemática de la diversidad de nomenclaturas asociadas a cada estación.
3. Las bases de datos carecen de la columna con las claves de las estaciones, con excepción del archivo `datos_2021`.
4. Algunos archivos contienen mediciones no correspondientes al año de monitoreo o, incluso, hojas de cálculo que no se desean procesar porque están vacías o los datos no son de interés (explicación de nomenclaturas).
5. El archivo `datos_2014` contiene un procesamiento y generación de estadísticos que deben eliminarse .
6. La columna relacionada a las horas del archivo `datos_2021` no está en formato adecuado.
7. Algunas hojas de `datos_2016` no tienen título en las columnas de fechas y horas de monitoreo.

La metodología de homogenización se presenta a continuación:
1. `pd.ExcelFile()` carga el archivo Excel y obtiene las hojas disponibles.
2. `sheet_names` detecta automáticamente las hojas de cálculo que se refieren a las estaciones de monitoreo.
3. `empty` verifica si una hoja está vacía antes de procesarla.
4. `columns.empty` revisa si la hoja tiene columnas o está vacía.
5. `rename()` modifica los nombres de columnas para uniformarlas.
6. `apply()`  aplica funciones para eliminar estadísticos de `datos_2014`.
7. `str.contains()` filtra columnas según palabras clave específicas.
8. `pd.to_datetime()` convierte la columna de fechas (`FECHA`) al formato datetime.
9. `dropna()` elimina filas con valores vacíos o NaT en la columna de fechas.
10. `dt.year` extrae el año de cada fila para el filtrado.
11. `concat()` une todas las hojas procesadas en un solo DataFrame.
12. `pd.to_timedelta()` convierte una columna a formato de duración/tiempo.
13. `replace()` sustituye los nombres de columnas dentro del DataFrame.
14. `append()` acumula las hojas de cálculo procesadas en una lista para luego fusionarlas.


```python
# Procesar los archivos de 2012 a 2018 y 2021
for año in [2012, 2013, 2014, 2015, 2016, 2017, 2018, 2021]:
    print(f"Procesando: datos_{año}.xlsx")

    # Cargar archivo Excel y obtener hojas (estaciones)
    ruta_archivo = f'./Archivos/Datos/datos_{año}.xlsx'
    df_excel = pd.ExcelFile(ruta_archivo)
    hojas = df_excel.sheet_names  # Detectar estaciones

    # Lista para acumular los datos de todas las hojas
    df_total = []

    # Procesar cada hoja (estación)
    for hoja in hojas:
        print(f"   Procesando estación: {hoja}")

        # Leer cada hoja
        df = pd.read_excel(ruta_archivo, sheet_name=hoja)

        # Si la hoja no tiene columnas ni datos, se omite
        if df.empty or df.columns.empty:
            print(f"Advertencia: La hoja '{hoja}' en '{ruta_archivo}' está vacía y fue omitida.")
            continue  # Saltar a la siguiente hoja

        # Si la hoja es "Nomenclatura", saltarla
        if hoja == "Nomenclatura":
            print(f" Hoja '{hoja}' omitida en el archivo {ruta_archivo}.")
            continue  # Saltar a la siguiente hoja sin procesarla

        
        # Agregar nombre de estación
        df['ESTACION'] = hoja

        # Asignar nombres a las primeras columnas de 2016 que no tienen el título de FECHA y HORA
        if año == 2016:
            if hoja in ["Oblatos", "Las Pintas"]:
                df.columns = ['FECHA', 'HORA'] + df.columns[2:].tolist()  # Mantener el resto de columnas

        # Homogeneizar nombres de columnas
        columnas_homogeneas = {'Date': 'FECHA',
                               'Fecha': 'FECHA',
                               'Time': 'HORA',
                               'Hora': 'HORA',
                               'Santa Fé' : 'FECHA',
                               'Fecha ': 'FECHA', 
                               'Hora ': 'HORA', 
                               '1:O3': 'O3',
                               '17:O3': 'O3',
                               'Ozono': 'O3',
                               '18:NO': 'NO',
                               '3:NO2': 'NO2',
                               '19:NO2': 'NO2',
                               '20:NOx': 'NOx',
                               'Nox': 'NOx',
                               'NOX': 'NOx',
                               '5:SO2': 'SO2',
                               '21:SO2': 'SO2',
                               '6:CO': 'CO',
                               '22:CO': 'CO',
                               '7:PM10': 'PM10',
                              '7:PM 10': 'PM10',
                               'PM 10': 'PM10',
                              'PM 2.5': 'PM2.5',
                              'Pm2.5': 'PM2.5',
                              'Temp Int': 'TMPI',
                              'TempInt': 'TMPI',
                              'Temp Ext': 'TMP',
                              '9:Temp Ext': 'TMP',
                               'Temperatura exterior': 'TMP',
                               'TempExt': 'TMP',
                               'TEM EXT': 'TMP',
                               'Temperatura': 'TMP',
                               'Temperatura ': 'TMP',
                               'HR': 'RH',
                               '10:RH': 'RH',
                               'Humedad relativa': 'RH',
                               'Humedad relativa ': 'RH',
                               '11:WS': 'WS',
                               'Velocidad de viento': 'WS',
                               'Velocidad de viento ': 'WS',
                               '12:WD': 'WD',
                               'Dirección de vientos': 'WD',
                               'Dirección de vientos ': 'WD',
                               'Precipitación': 'PP',
                               'Precipitacion': 'PP',
                               'Pr': 'PP',
                               'Precipíta ': 'PP',
                               'Precipita ': 'PP',
                               'Precipíta': 'PP',
                               'Precipita': 'PP',                               
                               'Radiación solar': 'RS',
                               'Radiación solar ': 'RS',
                               'Radiacion Solar': 'RS',
                               'RadSolar': 'RS',
                               'Rad Solar': 'RS',
                               'Presión': 'PBA',
                               'Presion': 'PBA',
                               'índice UV': 'UVI',
                               'Índice UV': 'UVI'
                              }
        
        df.rename(columns=columnas_homogeneas, inplace=True)

        # Eliminar filas innecesarias del archivo de 2014
        if año == 2014:
            df = df[~df.apply(lambda row: row.astype(str).str.contains("SUMA|PROMEDIO|Mínima|Máxima").any(), axis=1)]

        if 'FECHA' in df.columns:
            df['FECHA'] = pd.to_datetime(df['FECHA'], errors='coerce')
            df = df.dropna(subset=['FECHA'])  # Eliminar filas con NaT
            df = df[df['FECHA'].dt.year == año]
        else:
            print(f"Advertencia: La hoja '{hoja}' en '{ruta_archivo}' no tiene columna FECHA. Se omitió el filtrado por año.")

        # Filtrar registros que no corresponden al año del archivo
        df = df[df['FECHA'].dt.year == año]

        # Lista de columnas a eliminar
        columns_to_drop = ['Unnamed: 14', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Pintas', 'Estación SIMAJ']
        # Eliminar columnas solo si existen en el DataFrame
        df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])
                
        # Formatear HORA a hh:mm:ss si es numérica
        if df['HORA'].dtype in ['int64', 'float64']:  # Revisar si la columna es numérica
            df['HORA'] = pd.to_timedelta(df['HORA'], unit='h').astype(str).str.replace("0 days ", "")

        # Mover ESTACION a la primera columna
        cols = ['ESTACION'] + [col for col in df.columns if col != 'ESTACION']
        df = df[cols]

        # Acumular cada hoja en la lista
        df_total.append(df)

    # Concatenar todas las hojas del archivo en un solo DataFrame
    df_final = pd.concat(df_total, ignore_index=True)

    # Guardar cada archivo procesado en la carpeta
    df_final.to_excel(f'{carpeta_salida}/datos_{año}_homogeneizado.xlsx', index=False)

    print(f"Proceso completado: Guardado como {carpeta_salida}/datos_{año}_homogeneizado.xlsx")
```

Este proceso de homogenización se realizó por cada base de datos, guardando los archivos procesados con el nombre de `datos_{año}_homogeneizado`. Estos nuevos archivos se concatenaron mediante el método `.append.()` y `.concat()`. El dataframe final comprende datos desde **2011** a **2018**, incluyendo **2021**. 

```python
# Inicialización de la Lista
archivos = []

# Asignación de los Archivos a la Lista
for año in [2012, 2013, 2014, 2015, 2016, 2017, 2018, 2021]:
    df = pd.read_excel(f'./Archivos/Datos/Homogeneizados/datos_{año}_homogeneizado.xlsx')

    # Filtrar registros que no corresponden al año del archivo
    df = df[df['FECHA'].dt.year == año]
    df['HORA'] = pd.to_datetime(df['HORA'], format='%H:%M:%S', errors='coerce').dt.time

    archivos.append(df)  # Agregar el DataFrame filtrado a la lista
    
# Concatenación de los Elementos de la Lista a un DataFrame
df_12_18_21 = pd.concat(archivos, ignore_index = True)

# Guardar el DataFrame final como CSV
ruta_guardado = "./Archivos/Datos/Homogeneizados/datos_2012_2018_2021_homogeneizado.csv"
df_12_18_21.to_csv(ruta_guardado, index=False, encoding="utf-8")

# Mostrar el tipo de datos contenidos en el Dataframe
print('Tipos de Datos en el DataFrame:\n', df_12_18_21.dtypes)

# Impresión de los Resultados
display(df_12_18_21)
```

### Periodo 2019-2023
Para homogeneizar los datos monitoreo de la calidad del aire desde **2019** a **2023** se deben tener en cuenta los siguientes puntos: 
1. Aunque todos los archivos contienen los datos de interés en una única hoja de cálculo, hay bases de datos con hojas adicionales que no serán procesadas.
2. Se deberán homogeneizar los nombres de columnas (parámetros de medición).
3. Algunos archivos contienen mediciones no correspondientes al año de monitoreo o, incluso, hojas de cálculo que no se desean procesar porque están vacías o los datos no son de interés (explicación de nomenclaturas).

En este sentido, la metodología de homogenización fue similar al procesamiento del periodo anterior de bases de datos. 


```python
# Procesar los archivos de 2019 a 2023, uno por uno
for año in [2019, 2020, 2022, 2023]:
    print(f"Procesando: datos_{año}.xlsx")  # Mensaje de seguimiento

    # Cargar archivo Excel y obtener hojas (datos)
    ruta_archivo = f'./Archivos/Datos/datos_{año}.xlsx'
    df_excel = pd.ExcelFile(ruta_archivo)
    hojas = df_excel.sheet_names  # Detectar hojas con datos

    # Lista para acumular los datos de todas las hojas
    df_total = []

    # Procesar cada hoja (datos)
    for hoja in hojas:
        print(f"   Procesando estación: {hoja}")

        # Leer cada hoja
        df = pd.read_excel(ruta_archivo, sheet_name=hoja)

        # Si la hoja no tiene columnas ni datos, se omite
        if df.empty or df.columns.empty:
            print(f"Advertencia: La hoja '{hoja}' en '{ruta_archivo}' está vacía y fue omitida.")
            continue  # Saltar a la siguiente hoja

        # Omitir hojas sin datos de interes
        hojas_excluir = ['Nomenclatura-Unidad de medida ', 'Estaciones-Ubicación', 'unidades y parámetros']

        if hoja in hojas_excluir:
            print(f" Hoja '{hoja}' omitida en el archivo {ruta_archivo}.")
            continue  # Saltar a la siguiente hoja sin procesarla

        # Homogeneizar nombres de columnas
        columnas_homogeneas = {'Date_Time': 'FECHA',
                               'Fecha': 'FECHA',
                               'FECHA ': 'FECHA',
                               'Hora': 'HORA',
                               'HORA ': 'HORA',
                               'Estación': 'ESTACIÓN',
                               'CO ': 'CO',
                               'NO2 ': 'NO2',
                               'NOX': 'NOx',
                               'NOX ': 'NOx',
                               'PM10 ': 'PM10',
                               'PM 2.5': 'PM2.5',
                               'Temp Int': 'TMPI',
                               'Temp Int ': 'TMPI',
                               'Temp Ext': 'TMP',
                               'Precipitación': 'PP',
                               'PRECIPITACION': 'PP',
                               'PRECIPITACION ': 'PP',
                               'RH ': 'RH',
                               'Radiación Solar': 'RS',
                               'Radiación Solar ': 'RS',
                               'Rad Solar': 'RS',
                               'Presión Barometrica': 'PBA',
                               'PRESION BAROMETRICA': 'PBA',
                               'PRESION BAROMETRICA ': 'PBA',
                              }
        
        df.rename(columns=columnas_homogeneas, inplace=True)

        # Filtrar registros que no corresponden al año del archivo
        df = df[df['FECHA'].dt.year == año]
                
        # Formatear HORA a hh:mm:ss si es numérica
        if df['HORA'].dtype in ['int64', 'float64']:  # Revisar si la columna es numérica
            df['HORA'] = pd.to_timedelta(df['HORA'], unit='h').astype(str).str.replace("0 days ", "")

        # Mover ESTACION a la primera columna
        cols = ['ESTACIÓN'] + [col for col in df.columns if col != 'ESTACIÓN']
        df = df[cols]

        # Acumular cada hoja en la lista
        df_total.append(df)

    # Concatenar todas las hojas del archivo en un solo DataFrame
    df_final = pd.concat(df_total, ignore_index=True)

    # Guardar cada archivo procesado en la carpeta
    df_final.to_excel(f'{carpeta_salida}/datos_{año}_homogeneizado.xlsx', index=False)

    print(f"Proceso completado: Guardado como {carpeta_salida}/datos_{año}_homogeneizado.xlsx")
```

Este proceso de homogenización se realizó por cada base de datos, guardando los archivos procesados con el nombre de `datos_{año}_homogeneizado`. Estos nuevos archivos se concatenaron mediante el método `.append.()` y `.concat()`. El dataframe final comprende datos desde **2019** a **2023**.
